```python
# Data manipulation
import numpy as np
import pandas as pd

# Preprocessing
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Models
from sklearn.linear_model import Ridge
from xgboost import XGBRegressor

# Model selection and evaluation
from sklearn.model_selection import KFold, cross_val_score, GridSearchCV
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.base import clone
# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

1) Load the kNN data

```python
# Load KNN-imputed features and the target (aligned on the 1152 kept patients)
X_train_knn = pd.read_csv("../X_train_knn.csv", index_col="id")
X_test_knn = pd.read_csv("../X_test_knn.csv", index_col="id")
y_train_knn = pd.read_csv("../y_train.csv", index_col="id")["y"].loc[X_train_knn.index]

# 5-fold cross-validation (same split as Stacking/folds.csv)
cv = KFold(n_splits=5, shuffle=True, random_state=42)

# Check consistency
print("KNN shape:", X_train_knn.shape)
print("Same features in train and test:", X_train_knn.columns.equals(X_test_knn.columns))
print("Labels aligned:", X_train_knn.index.equals(y_train_knn.index))
```

2) Ridge regression - standard scaler and alpha tuning

```python
# Pipeline: Scaling + Ridge Regression
ridge_knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("ridge", Ridge())
])

# Hyperparameter grid
ridge_knn_param_grid = {
    "ridge__alpha": np.logspace(-3, 5, 30)
}

# Grid search with the same CV folds
ridge_knn_search = GridSearchCV(
    estimator=ridge_knn_pipeline,
    param_grid=ridge_knn_param_grid,
    scoring="r2",
    cv=cv,
    n_jobs=-1
)

ridge_knn_search.fit(X_train_knn, y_train_knn)

# Results
print("Best alpha:", ridge_knn_search.best_params_["ridge__alpha"])
print("Best CV R²:", ridge_knn_search.best_score_)
print("Training R²:", ridge_knn_search.score(X_train_knn, y_train_knn))
```

3) XGBoost 
Tuning of : - n_estimtors
            - max_depth
            - learning_rate
            - min_child_weight (contrôle les divisions des arbres pour limiter l'overfitting)
On utilise le même cv (même Kfold que pour ridge)

```python
# XGBoost model
xgb_knn = XGBRegressor(
    objective="reg:squarederror",
    random_state=42,
    n_jobs=1,
    tree_method="hist"
)

# Hyperparameter grid
xgb_knn_param_grid = {
    "n_estimators": [200, 500],
    "max_depth": [2, 3, 5],
    "learning_rate": [0.03, 0.1],
    "min_child_weight": [1, 5]
}

# Grid search with the same CV folds
xgb_knn_search = GridSearchCV(
    estimator=xgb_knn,
    param_grid=xgb_knn_param_grid,
    scoring="r2",
    cv=cv,
    n_jobs=-1
)

xgb_knn_search.fit(X_train_knn, y_train_knn)

# Results
print("Best parameters:", xgb_knn_search.best_params_)
print("Best CV R²:", xgb_knn_search.best_score_)
print("Training R²:", xgb_knn_search.score(X_train_knn, y_train_knn))
```

Keep best param from above and add early stopping to try and limit overfit 

```python
# Store results
early_stop_scores = []
best_iterations = []
train_scores = []

for train_idx, val_idx in cv.split(X_train_knn):

    # Outer training and validation folds
    X_tr = X_train_knn.iloc[train_idx]
    X_val = X_train_knn.iloc[val_idx]

    y_tr = y_train_knn.iloc[train_idx]
    y_val = y_train_knn.iloc[val_idx]

    # Inner split for early stopping
    X_fit, X_es, y_fit, y_es = train_test_split(
        X_tr, y_tr,
        test_size=0.2,
        random_state=42
    )

    # XGBoost model
    model = XGBRegressor(
        objective="reg:squarederror",
        n_estimators=2000,
        learning_rate=0.1,
        max_depth=3,
        min_child_weight=5,
        early_stopping_rounds=50,
        eval_metric="rmse",
        tree_method="hist",
        random_state=42,
        n_jobs=1
    )

    # Fit with early stopping
    model.fit(
        X_fit, y_fit,
        eval_set=[(X_es, y_es)],
        verbose=False
    )

    # Evaluate on outer validation fold
    y_pred = model.predict(X_val)

    early_stop_scores.append(r2_score(y_val, y_pred))
    best_iterations.append(model.best_iteration + 1)
    train_scores.append(r2_score(y_fit, model.predict(X_fit)))

# Results
print("Mean CV R²:", np.mean(early_stop_scores))
print("Std CV R²:", np.std(early_stop_scores))
print("Mean Training R²:", np.mean(train_scores))
print("Best number of trees per fold:", best_iterations)
print("Mean best number of trees:", np.mean(best_iterations))
```

We now keep the best parameters from above and tune the following to try and improve R2 in validation and reduce the overfitting : 
    - subsample 
    - colsample_bytree
    - reg_lambda
    - reg_alpha

```python
# XGBoost with best parameters from first KNN search
xgb_knn_reg = XGBRegressor(
    objective="reg:squarederror",
    learning_rate=0.1,
    max_depth=3,
    min_child_weight=5,
    n_estimators=500,
    tree_method="hist",
    random_state=42,
    n_jobs=1
)

# Regularization hyperparameter grid
knn_reg_param_grid = {
    "subsample": [0.6, 0.8, 1.0],
    "colsample_bytree": [0.6, 0.8, 1.0],
    "reg_lambda": [1, 10, 100],
    "reg_alpha": [0, 1, 10]
}

# Grid search with the same CV folds
xgb_knn_reg_search = GridSearchCV(
    estimator=xgb_knn_reg,
    param_grid=knn_reg_param_grid,
    scoring="r2",
    cv=cv,
    n_jobs=-1
)

xgb_knn_reg_search.fit(X_train_knn, y_train_knn)

# Results
print("Best parameters:", xgb_knn_reg_search.best_params_)
print("Best CV R²:", xgb_knn_reg_search.best_score_)
print("Training R²:", xgb_knn_reg_search.score(X_train_knn, y_train_knn))
```

Create submission 


```python
# Predict on the KNN-imputed test set
ridge_predictions = ridge_knn_search.best_estimator_.predict(X_test_knn)
xgb_predictions = xgb_knn_reg_search.best_estimator_.predict(X_test_knn)

# Load Kaggle submission template
sample = pd.read_csv("../sample.csv")

# Verify dimensions
assert len(sample) == len(X_test_knn)
assert sample.columns.tolist() == ["id", "y"]

# Create Ridge submission
ridge_submission = sample.copy()
ridge_submission["y"] = ridge_predictions
ridge_submission.to_csv("submission_ridge_knn.csv", index=False)

# Create XGBoost submission
xgb_submission = sample.copy()
xgb_submission["y"] = xgb_predictions
xgb_submission.to_csv("submission_xgboost_knn.csv", index=False)

print("Ridge submission:", ridge_submission.shape)
print("XGBoost submission:", xgb_submission.shape)
print("Missing predictions:", xgb_submission["y"].isna().sum())
```

Out-of-fold predictions on the shared folds (for the stacking)

```python
from pathlib import Path
from sklearn.base import clone

# Load shared folds
stacking_dir = Path("../Stacking")

folds = pd.read_csv(
    stacking_dir / "folds.csv", index_col="id"
)["fold"].loc[X_train_knn.index]

# Best XGBoost model
estimator = xgb_knn_reg_search.best_estimator_

# Initialize OOF predictions
oof = pd.Series(
    np.nan, index=X_train_knn.index, name="xgb"
)

# Generate out-of-fold predictions
for f in sorted(folds.unique()):
    train = folds != f
    val = folds == f

    model = clone(estimator)
    model.fit(X_train_knn.loc[train], y_train_knn.loc[train])

    oof.loc[val] = model.predict(X_train_knn.loc[val])

assert oof.notna().all()

print(f"XGBoost OOF R²: {r2_score(y_train_knn, oof):.4f}")

# Save OOF predictions
oof.to_csv(stacking_dir / "oof_xgb.csv")

# Test predictions: submission_xgboost_knn.csv (written above), read by Stacking/stacking.ipynb
```
